```python
import yfinance as yf
import numpy as np
import pandas as pd

from datetime import datetime
```

```python
# Import
transaction_data = pd.read_excel("./Transaction Data.xlsx")

# Get tickers:
all_tickers = transaction_data['Ticker'].unique()

# Remove delisted asset (manually):
blackList = []
filt_tickers = [tick for tick in all_tickers if tick not in blackList]

print("There are {} active assets and they are:".format(len(filt_tickers)))
print(filt_tickers)
```

# PRICE HISTORY

```python
start_Date = '2021-01-01'
end_Date = datetime.today().strftime("%Y-%m-%d")

raw = yf.download(
    filt_tickers,
    start=start_Date,
    end=end_Date,
    auto_adjust=True,
    progress=False
)

price_data = (
    raw.swaplevel(0, 1, axis=1) # swap (Field, Ticker) to (Ticker,Field)
    .sort_index(axis=1)
    .stack(level=0)
    .rename_axis(["Date","Ticker"])
    .reset_index()
    .set_index(["Ticker", "Date"])
    .sort_index()
)

price_data
```
